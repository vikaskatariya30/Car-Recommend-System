```python
df[df['mileage_or_range'] > 35][['name', 'fuel_type', 'mileage_or_range', 'mileage_range_bucket']]
```

```python
def bucket_mileage(row):
    val = row["mileage_or_range"]
    if row["fuel_type"] == "Electric":
        bins = [0, 200, 300, 400, 500, 600, 800]
        labels = ["0-200", "200-300", "300-400", "400-500", "500-600", "600-800"]
    else:
        bins = [0, 10, 15, 20, 25, 30, 35]
        labels = ["0-10", "10-15", "15-20", "20-25", "25-30", "30-35"]
    return pd.cut([val], bins=bins, labels=labels)[0]

df["mileage_range_bucket"] = df.apply(bucket_mileage, axis=1)
```
